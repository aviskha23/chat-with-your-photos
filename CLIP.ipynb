```python
from datasets import load_dataset

dataset = load_dataset(
    "nlphuji/flickr30k",
    revision="refs/convert/parquet",
    split="test"
)

print(dataset)
```

```python
print(len(dataset))
print(dataset.column_names)
```

```python
NUM_IMAGES = 500

subset = dataset.select(range(NUM_IMAGES))

print(len(subset))
```

```python

import os
import zipfile
from google.colab import files

# Create a folder for the actual image files
os.makedirs("images", exist_ok=True)

# Keep filenames in exactly the same order as the dataset records
filenames = [
    os.path.basename(subset[i]["filename"])
    for i in range(len(subset))
]

# Save all 500 images
for i in range(len(subset)):
    image = subset[i]["image"].convert("RGB")
    image.save(os.path.join("images", filenames[i]))

    if (i + 1) % 50 == 0:
        print(f"Saved {i + 1}/500 images")

# Package the images into a ZIP file
with zipfile.ZipFile(
    "images.zip",
    "w",
    compression=zipfile.ZIP_DEFLATED
) as zip_file:
    for filename in filenames:
        image_path = os.path.join("images", filename)
        zip_file.write(
            image_path,
            arcname=f"images/{filename}"
        )

print("ZIP created successfully!")
print("Number of images:", len(filenames))

# Download the ZIP to your laptop
files.download("images.zip")
```
